# Funcion 

- realiza una función llamada agregar_una_vez: que reciba una lista y un elemento, La funcion debe añadir el elemento al final de lista con la condición de no repetir ningun elemento,+
ademas si este elemento ya se encuentra en la lista se debe invocar un error de tipo ValueError, que debes capturar y mostrar el mensaje en su lugar:


```python
elementos = [1, 5, -2]
def agregar_una_vez(lista, elemento):
    try:
        if elemento in lista:
            raise ValueError
        else:
            lista.append(elemento)
    except ValueError:
        print("El elemento ya existe en la lista {}".format(elemento))

agregar_una_vez(elementos, 2)
agregar_una_vez(elementos, 10)
agregar_una_vez(elementos, 5)
print(elementos)
```
